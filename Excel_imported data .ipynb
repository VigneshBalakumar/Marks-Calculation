```python
import matplotlib.pyplot as plt

# Sri's marks
subjects = ['English', 'Tamil', 'Science', 'Computer Science', 'Maths', 'Social']
marks = [50, 70, 60, 70, 64, 56]

# Bar chart
plt.figure(figsize=(8,4))
plt.bar(subjects, marks, color='yellow')
plt.title("Sri's Subject-wise Performance")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.ylim(0,100)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
from scipy.stats import ttest_1samp

data = [75, 82, 88, 78, 95, 89, 92, 85, 88, 79]
population_mean = 85

t_stat, p_val = ttest_1samp(data, population_mean)
print(f"T-statistic: {t_stat}, P-value: {p_val}")
```

```python
from scipy.stats import ttest_ind

group1 = [64, 58, 66, 75, 57, 57, 75, 67, 55, 65]
group2 = [80, 80, 87, 65, 67, 79, 74, 88, 75, 70]

t_stat, p_val = ttest_ind(group1, group2)
print(f"T-statistic: {t_stat}, P-value: {p_val}")
```
